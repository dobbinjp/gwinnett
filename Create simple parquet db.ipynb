```python
folderloc = "abfss://raw@gwisynadls.dfs.core.windows.net/saleslt/"
dbname = "adventureworks_parquet"
```

```python
spark.sql(f"create database if not exists {dbname};")
```

```python
spark.sql(f"use database {dbname};")
```

```python
for f in notebookutils.fs.ls(folderloc):
    print('processing ' + f.name + '...')
    table_name = f.name
    df = spark.read.format("parquet").load(f.path)
    df.write.format("parquet").saveAsTable(table_name)
print('processing complete!')
```
